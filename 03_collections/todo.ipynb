```python
filename = "todo_list.txt"

# ファイル読み込み（空行は除外）
try:
    with open(filename, "r", encoding="utf-8") as f:
        todo_list = [line.strip() for line in f if line.strip()]
except FileNotFoundError:
    todo_list = []

print("【TODOリスト】")

while True:
    raw = input("番号を入力してください[1:追加、2:表示、3:削除、4:保存と終了]:").strip()
    choice = raw[0] if raw else ""

    if choice == "1":
        print("番号を入力してください[1:追加、2:表示、3:削除、4:保存と終了]:1:追加")
        add_todo = input("追加するTODO: ")
        todo_list.append(add_todo)

    elif choice == "2":
        print("番号を入力してください[1:追加、2:表示、3:削除、4:保存と終了]:2:表示")
        for i, todo in enumerate(todo_list, start=1):
            print(f"{i}. {todo}")

    elif choice == "3":
        print("番号を入力してください[1:追加、2:表示、3:削除、4:保存と終了]:3:削除")
        for i, todo in enumerate(todo_list, start=1):
            print(f"{i}. {todo}")

        # ★
        delete_input = input("削除する番号: ").strip()
        print(f" 削除する番号: {delete_input} ")

        try:
            delete_no = int(delete_input)

            # 課題例に合わせる：1を入力したら2つ目が消える（0始まり扱い）
            delete_index = delete_no

            if 0 <= delete_index < len(todo_list):
                todo_list.pop(delete_index)
            else:
                print("無効な番号です。")
        except ValueError:
            print("無効な番号です。")

    elif choice == "4":
        print("番号を入力してください[1:追加、2:表示、3:削除、4:保存と終了]:4:保存と終了")
        with open(filename, "w", encoding="utf-8") as f:
            for todo in todo_list:
                f.write(todo + "\n")
        print("TODOリストを保存して終了します。")
        break

    else:
        print("1-4の数字を入力してください。")
```
