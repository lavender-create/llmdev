```python
# 成績を入力（カンマ区切り）
input_data = input("成績をカンマ区切りで入力してください（例: 85,90,78,92,88）：")

# 文字列 → 数値リストに変換
grades = [int(x) for x in input_data.split(",")]

# 入力された成績リストを表示
print(f"成績リスト: {grades}")

# 平均点
average = sum(grades) / len(grades)

# 最高点・最低点
max_score = max(grades)
min_score = min(grades)

# 平均以上の学生数
count_above_average = 0
for score in grades:
    if score >= average:
        count_above_average += 1

# 結果表示
print(f"平均点: {average}")
print(f"最高点: {max_score}")
print(f"最低点: {min_score}")
print(f"平均以上の学生数: {count_above_average}")
```
