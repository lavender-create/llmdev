```python
# 英単語を格納するための空のリスト
words = []

# アルファベットごとの数を格納する辞書（a〜z を 0 で初期化）
alphabet_count = {
    'a': 0, 'b': 0, 'c': 0, 'd': 0, 'e': 0,
    'f': 0, 'g': 0, 'h': 0, 'i': 0, 'j': 0,
    'k': 0, 'l': 0, 'm': 0, 'n': 0, 'o': 0,
    'p': 0, 'q': 0, 'r': 0, 's': 0, 't': 0,
    'u': 0, 'v': 0, 'w': 0, 'x': 0, 'y': 0,
    'z': 0
}

# 英単語の入力を繰り返す
while True:
    word = input("英単語を入力してください（終了するには Enter）：")

    # 入力が空文字なら終了
    if word == "":
        break

    # 英単語をリストに追加
    words.append(word)

    # 英単語を1文字ずつ確認
    for char in word:
        # 小文字に変換して a〜z の場合のみカウント
        char = char.lower()
        if char in alphabet_count:
            alphabet_count[char] += 1


# 英単語リストを表示
for w in words:
    print(f"英単語を入力してください：{w}")

# 入力した英単語を表示
# ABC順に並べ替えて表示
words.sort()
print(f"\n入力した英単語：{words}")

# アルファベットごとの出現回数を表示

for key, value in alphabet_count.items():
    if value > 0:
        print(f"{key} が {value} 個ありました")
```
