```python
#税抜価格が2000円の場合
price = float(input("税抜価格を入力してください："))

tax_rate = 0.10
tax_included_price = price * (1 + tax_rate)

if tax_included_price >= 2000:
    shipping = 0
    print("送料は 無料です")
else:
    shipping = 350
    print(f"送料込価格は {int(350)} 円です")

total_price = tax_included_price + shipping


print(f"送料込価格は {int(total_price)} 円です")
```
