```python
# 元金（整数）の入力
principal = int(input("元金を入力してください："))
print(f"元金を入力してください：{principal} 円")

# 年利（％・整数）の入力
rate_percent = int(input("年利（％）を入力してください："))
print(f"年利を入力してください：{rate_percent} %")

# 年利を小数に変換
rate = rate_percent / 100

# 複利計算（10年間）
amount = principal
for _ in range(10):
    amount *= (1 + rate)

# 結果の表示（小数第2位まで）
print(f"10年後の金額は {amount:.2f} 円です")
# 金利計算の課題完了


```
