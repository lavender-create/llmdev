```python
try:
    # 入力
    a = int(input("1つ目の数値を入力してください："))
    b = int(input("2つ目の数値を入力してください："))

    # 入力値を表示
    print(f"入力された数値1：{a}")
    print(f"入力された数値2：{b}")

    # 割り算
    result = a / b
    print(f"計算結果: {result} です")

except ZeroDivisionError:
    print("エラー: 0で割ることはできません")

except ValueError:
    print("エラー: 数値を入力してください")

    

```
